## Count Records

```python
SPLITS = [
    "train_formula_direct",
    "val_formula_direct",
    "train_algorithmic_scaffold",
    "val_algorithmic_scaffold",
    "train_algorithmic_scaffold_v2",
    "val_algorithmic_scaffold_v2",
    "train_algorithmic_scaffold_v2_5",
    "val_algorithmic_scaffold_v2_5",
    "train_algorithmic_scaffold_v3",
    "val_algorithmic_scaffold_v3",
    "train_algorithmic_scaffold_v3_1",
    "val_algorithmic_scaffold_v3_1",
    "train_algorithmic_scaffold_v3_5",
    "val_algorithmic_scaffold_v3_5",
]
```

```python
rows = []

for split in SPLITS:
    split_dir = PROJECT_ROOT / "benchmark" / "data" / split
    for path in sorted(split_dir.glob("*.jsonl")):
        count = 0
        answer_types = {}
        for line in path.read_text().splitlines():
            if not line.strip():
                continue
            record = json.loads(line)
            count += 1
            answer_type = record["answer_type"]
            answer_types[answer_type] = answer_types.get(answer_type, 0) + 1
        rows.append({
            "split": split,
            "file": path.name,
            "count": count,
            "answer_types": answer_types,
        })

rows
```
