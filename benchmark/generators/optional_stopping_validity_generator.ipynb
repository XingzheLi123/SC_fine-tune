# Optional-Stopping Validity Generator

Notebook workbench for generating prompts, reasoning traces, and canonical answers for optional-stopping validity problems.

```python
from pathlib import Path
import json
import sys

project_root = Path.cwd()
while project_root != project_root.parent and not (project_root / "benchmark").exists():
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
```

```python
from benchmark.generators import OptionalStoppingValidityGenerator

gen = OptionalStoppingValidityGenerator()
```

```python
params = gen.sample_params(seed=201, difficulty=2, split="dev")
params
```

```python
problem = gen.generate_problem(params)
reasoning = gen.generate_reasoning(params)
solution = gen.generate_solution(params)

print("PROBLEM:\n", problem)
print("\nREASONING:\n", reasoning)
print("\nSOLUTION:\n", gen.to_json_safe(solution))
```

```python
records = []
for difficulty in [1, 2, 3]:
    for seed in range(2000 + 100 * difficulty, 2005 + 100 * difficulty):
        records.append(gen.generate_record(seed=seed, difficulty=difficulty, split="dev"))

len(records), records[0]
```

```python
output_path = project_root / "benchmark" / "data" / "dev" / "optional_stopping_validity_preview.jsonl"
with output_path.open("w") as f:
    for record in records:
        f.write(json.dumps(record, sort_keys=True) + "\n")

output_path
```
